## Null Values Using Heat-Map

```python
sns.heatmap(c.isnull())
```

```python
#Filling Missing Categorical values
c['director'].fillna('unknown' , inplace =True)
c['cast'].fillna('unknown' , inplace =True)
c['country'].fillna('unknown' , inplace =True)
```

```python
c.isnull().sum()
```

```python
#Dropping Rows
c.dropna(subset = ['date_added','rating'],inplace =True)
```

```python
c.isnull().sum()
```

```python
# clean date format
c['date_added'] = pd.to_datetime(c['date_added'].str.strip())
```

```python
c.dtypes
```

```python
c.info()
```

```python
# Extract Year and Month Added
c['year_added']=c['date_added'].dt.year
c['month_added'] = c['date_added'].dt.month_name()
```

```python
c.head()
```

## Unique Values in Each column

```python
for i in c.columns:
    print(i,':','\n',c[i].unique())
```

```python
a=c.describe(include='all')
a
```

```python
for i in ['director','cast','country','date_added','rating']:
    c[i].replace(np.nan,a[i][2],inplace=True)
```

## Separating Movies & TV Shows by creating a new variable

```python
movie_countries=c[c['type']=='Movie']
tv_show_countries=c[c['type']=='TV Show']
```

```python
plt.figure(figsize=(12,6))
sns.countplot(y='country' , order=c.country.value_counts().index[0:10] , data=movie_countries)
plt.title('Top 10 countries producing Movies in Netflix')

plt.figure(figsize=(12,6))
sns.countplot(y='country' , order=c.country.value_counts().index[0:10] , data=tv_show_countries)
plt.title('Top 10 countries producing TV Show in Netflix')
```
